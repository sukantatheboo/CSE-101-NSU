## Problem 10

Write a function `clamp(x, low, high)` that returns:
- `low` if `x < low`
- `high` if `x > high`
- otherwise `x`
Print results for `clamp(5, 1, 10)`, `clamp(-2, 0, 3)`, `clamp(99, 0, 50)`.

```python
# TODO: Define clamp(x, low, high)

# TODO: Print clamp(5, 1, 10)
# TODO: Print clamp(-2, 0, 3)
# TODO: Print clamp(99, 0, 50)

def clamp(x, low, high):
    if x < low:
        print(low)
    elif x > high:
        print(high) 
    else:
        print(x)

clamp(5, 1, 10)
clamp(-2, 0, 3)
clamp(99, 0, 50)
```

## Problem 15

Repeatedly read integers until the user enters `done`.
Ignore any input that is not a valid integer (use `try/except` and `continue`).
At the end, print the total sum of valid integers.

```python
# TODO: Loop and read input
# TODO: Break on 'done'
# TODO: Try converting to int; if fails, continue
# TODO: Add to total
# TODO: Print total

total = 0

while True:
    line = input('> ')

    if line == 'done':
        break
    
    try:
        num = int(line)
        total = total + num 
    except ValueError:
        continue

print(total)
```

## Problem 24

Write a function `count_vowels(s)` that returns how many vowels are in `s` (a, e, i, o, u). Print the result for `'education'`.

```python
# TODO: Define count_vowels(s)

# TODO: Print count_vowels('education')

def count_vowels(s):
    vowels ='aeiou'
    count = 0
    
    for char in s:
        if char in vowels:
            count += 1
    return count
print(count_vowels('education'))
```

## Problem 25

Write a function `summarize(nums)` that returns a tuple:
`(count, total, average, smallest, largest)`
Constraints: no `len()`, `sum()`, `min()`, `max()`.
Test with `data = [12, 3, 19, 7, 7, 25]` and print the returned tuple.

```python
# TODO: Define summarize(nums)
# TODO: Compute count, total, smallest, largest using loop idioms
# TODO: Compute average after loop
# TODO: Return (count, total, average, smallest, largest)
# TODO: Print summarize(data)

def summarize(nums):
    count = 0
    total = 0
    smallest = nums[0]
    largest = nums[0]
    
    for num in nums:
        count += 1
        total += num
        
        if num < smallest:
            smallest = num
            
        if num > largest:
            largest = num
            
    average = total / count if count > 0 else 0
    
    return (count, total, average, smallest, largest)

data = [12, 3, 19, 7, 7, 25]
print(summarize(data))
```
