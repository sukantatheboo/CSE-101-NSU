## **Problem 1**

Ask the user to enter numbers until they type `"done"`.

Store the numbers in a list.

After input ends, print:

- The full list
- The largest number
- The smallest number
- The sum of the numbers
- The average of the numbers

````python
# Write your solution here
def process_numbers():
    numbers = []

    while True:
        user_input = input("Enter a number (or type 'done' to finish): ")
    
        if user_input.lower() == 'done':
            break
        
        try:
            num = float(user_input)
            numbers.append(num)
        except ValueError:
            print("Invalid input. Please enter a valid number.")

    if len(numbers) > 0:
        total_sum = sum(numbers)
        count = len(numbers)    
        print(f"The full list: {numbers}")
        print(f"The largest number: {max(numbers)}")
        print(f"The smallest number: {min(numbers)}")
        print(f"The sum of the numbers: {total_sum}")
        print(f"The average of the numbers: {total_sum / count}")
    else:
        print("No numbers were entered.")
        
process_numbers()
````

## **Problem 3**

Given the list:

```python
values = [12, 5, 8, 21, 3, 18, 10, 7, 16]
```

Create three new lists:

- numbers **less than 10**
- **even numbers**
- **odd numbers**

Print all three lists.

### Example Output
```python
Less than 10: [5, 8, 3, 7]
Even numbers: [12, 8, 18, 10, 16]
Odd numbers: [5, 21, 3, 7]
```


````python
# Write your solution here

values = [12, 5, 8, 21, 3, 18, 10, 7, 16]
less_than_10 = []
even_numbers = []
odd_numbers = []

for num in values:
    if num < 10:
        less_than_10.append(num)
    if num % 2 == 0:
        even_numbers.append(num)
    else:
        odd_numbers.append(num)

print(f"Less than 10: {less_than_10}")
print(f"Even numbers: {even_numbers}")
print(f"Odd numbers: {odd_numbers}")
````

## **Problem 8**

Given the list:

```python
names = ['ana', 'bob', 'ana', 'carl', 'bob', 'ana', 'dina', 'carl', 'ana']
```

Create a dictionary that counts how many times each name appears.

Then print:

- the dictionary
- the name with the **highest count**
- its count

### Example Output
```python
{'ana': 4, 'bob': 2, 'carl': 2, 'dina': 1}
Most frequent name: ana
Count: 4
```

````python
# Write your solution here
names = ['ana', 'bob', 'ana', 'carl', 'bob', 'ana', 'dina', 'carl', 'ana']

name_counts = {}
for name in names:
    if name in name_counts:
        name_counts[name] += 1
    else:
        name_counts[name] = 1

print(name_counts)

highest_name = ""
highest_count = 0

for name, count in name_counts.items():
    if count > highest_count:
        highest_name = name
        highest_count = count

print(f"Most frequent name: {highest_name}")
print(f"Count: {highest_count}")
````

Given the following text:

```python
'''
```

Write a program that:

1. Splits the text into individual words.
2. Creates a dictionary that counts how many times each word appears.

## **Problem 10**

Given the dictionary:

```python
scores = {
    'Alice': 85,
    'Bob': 91,
    'Charlie': 78,
    'Diana': 91,
    'Evan': 88,
    'Farah': 78
}
```

Find the **highest score** and print the names of **all students** who have that score.

### Example Output
```text
Highest score: 91
Students:
Bob
Diana
```

````python
# Write your solution here
scores = {
    'Alice': 85,
    'Bob': 91,
    'Charlie': 78,
    'Diana': 91,
    'Evan': 88,
    'Farah': 78
}

highest_score = max(scores.values())

print(f"Highest score: {highest_score}")
print("Students:")

for name, score in scores.items():
    if score == highest_score:
        print(name)
````

## **Problem 13**

Given the dictionary:

```python
prices = {
    'pen': 12,
    'book': 55,
    'bag': 320,
    'eraser': 8
}
```

Convert the dictionary into a list of tuples in the form:

```python
(value, key)
```

Sort the list in **reverse order** and print it.

### Example Output
```python
[(320, 'bag'), (55, 'book'), (12, 'pen'), (8, 'eraser')]
```

````python
# Write your solution here
prices = {
    'pen': 12,
    'book': 55,
    'bag': 320,
    'eraser': 8
}

price_list = []
for key, value in prices.items():
    price_list.append((value, key))

price_list.sort(reverse=True)
print(price_list)
````

# **Problem 16**

Given:

```python
scores = [
    ('Asha', 67),
    ('Borna', 91),
    ('Ritu', 84),
    ('Nafi', 91),
    ('Sami', 67),
    ('Lopa', 84)
]
```

Create a dictionary where:

- each **key** is a score
- each **value** is the number of students who got that score

Print the dictionary.

### Example Output
```python
{67: 2, 91: 2, 84: 2}
```

````python
# Write your solution here
scores = [
    ('Asha', 67),
    ('Borna', 91),
    ('Ritu', 84),
    ('Nafi', 91),
    ('Sami', 67),
    ('Lopa', 84)
]

score_counts = {}

for name, score in scores:
    if score in score_counts:
        score_counts[score] += 1
    else:
        score_counts[score] = 1

print(score_counts)
````
