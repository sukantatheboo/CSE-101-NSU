## 1. Attendance Tracker 
A teacher records student IDs who entered the class, but some students entered multiple times.

Return a list of **unique student IDs in the order they first appeared**.

**Example Input:**
[101, 102, 101, 103, 102, 104]

**Expected Output:**
[101, 102, 103, 104]

```python
# Write your solution here
inp =[101, 102, 101, 103, 102, 104]
exp_opt = []
for num in inp:
    if num not in exp_opt:
        exp_opt.append(num)
print(exp_opt)
```

## 2. Grocery Bill Counter
A grocery store records items sold in a day.

Create a dictionary that shows **how many times each item was sold**.

**Example Input:**
["milk", "bread", "milk", "eggs", "bread", "milk"]

**Expected Output:**
{"milk": 3, "bread": 2, "eggs": 1}

```python
# Write your solution here
inp =["milk", "bread", "milk", "eggs", "bread", "milk"]
opt ={}
for item in inp:
    if item in opt:
        opt[item] += 1
    else:
        opt[item] = 1
print(opt)
```

## 3. Temperature Report
A weather station stores daily temperatures in a tuple.

Return a tuple containing:
- highest temperature
- lowest temperature

**Example Input:**
(30, 32, 29, 35, 31)

**Expected Output:**
(35, 29)

```python
# Write your solution here
inp_temp = (30, 32, 29, 35, 31)
highest = inp_temp[0]
lowest = inp_temp[0]
for temp in inp_temp:
    if temp > highest:
        highest = temp
    elif temp < lowest:
        lowest = temp

opt_temp = (highest, lowest)
print(opt_temp)
```

## 4. Password Checker 
A password must contain at least one vowel.

Return "Valid" if it contains a vowel, otherwise "Invalid".

**Example Input:**
'try123'

**Expected Output:**
'Invalid'

```python
# Write your solution here
inp = input('Enter ur password')
vowels = 'aeiouAEIOU'
for char in inp:
    if char in vowels:
        print('valid')
        break
else:
    print('invalid')
```

## 5. Student Score Summary 
A teacher records marks as a list of tuples `(student_name, marks)`.

Create a dictionary where:
- key = student name
- value = total marks

**Example Input:**
[("Alice", 10), ("Bob", 15), ("Alice", 20), ("Bob", 5)]

**Expected Output:**
{"Alice": 30, "Bob": 20}

```python
# Write your solution here
inp = [("Alice", 10), ("Bob", 15), ("Alice", 20), ("Bob", 5)]
total_marks = {}
for name,mark in inp:
    if name in total_marks:
        total_marks[name] += mark
    else:
        total_marks[name] = mark
print(total_marks)
```
