## Problem 1 
Write a function `normalize_username(name)` that:
1) Strips leading/trailing spaces
2) Converts to lowercase
3) Replaces **all spaces** inside with underscores `_`
4) Returns the result

Example: `'  John Adam  '` → `'John Adam'`

```python
def normalize_username(name):
    return name.strip().lower().replace(' ', '_')

print(normalize_username('  John Adam  '))  
```

## Problem 2 — Count a Substring (Overlapping allowed)
Write a function `count_substring(text, sub)` that counts how many times `sub` appears in `text`, **including overlapping matches**.

Example: `count_substring('banana', 'ana')` should be `2` (positions 1 and 3).

```python
def count_substring(text, sub):
    count = 0
    start = 0
    while True:
        start = text.find(sub, start)
        if start == -1:
            break
        count += 1
        start += 1 
    return count

print(count_substring('banana', 'ana'))  
```

## Problem 3 — Title-Case Without `.title()`
Write a function `simple_title_case(sentence)` that:
- Strips extra spaces at the ends
- Splits words by spaces (assume single spaces between words after stripping)
- Capitalizes the first letter of each word and makes the rest lowercase
- Returns the new sentence

Example: `'hELLo wORLd'` → `'Hello World'`

```python
def simple_title_case(sentence):
    words = sentence.strip().split(' ')
    title_words = []
    
    for word in words:
        if len(word) > 0:
            new_word = word[0].upper() + word[1:].lower()
            title_words.append(new_word)
        else:
            title_words.append('')
            
    return ' '.join(title_words)

print(simple_title_case('hELLo wORLd'))
```

## Problem 4 — Mask Digits
Write a function `mask_digits(s)` that replaces every digit (`0`–`9`) with `*` and returns the result.

Example: `'Room 12B, Floor 7'` → `'Room **B, Floor *'`

```python
def mask_digits(s):
    result = ""
    for char in s:
        if char.isdigit():
            result += "*"
        else:
            result += char
    return result


print(mask_digits('Room 12B, Floor 7'))  
```

## Problem 5 — Replace Only the First Occurrence
Write a function `replace_first(text, old, new)` that replaces **only the first** occurrence of `old` with `new`.
If `old` is not found, return the original text.

Example: `replace_first('one one one', 'one', '1')` → `'1 one one'`

```python
def replace_first(text, old, new):
    return text.replace(old, new, 1)


print(replace_first('one one one', 'one', '1'))  
```

## Problem 6 — Parse and Sum Integers from a Mixed String
Write a function `sum_numbers_in_text(text)` that finds all **positive integers** embedded in `text` and returns their sum.

Rules:
- Numbers are sequences of digits.
- Use only string operations/loops.

Example: `'I ran 12 km and then 3 km'` → `15`

```python
def sum_numbers_in_text(text):
    current_num_str = ""
    total = 0
    
    for char in text:
        if char.isdigit():
            current_num_str += char
        else:
            if current_num_str != "":
                total += int(current_num_str)
                current_num_str = ""
                
    if current_num_str != "":
        total += int(current_num_str)
        
    return total


print(sum_numbers_in_text('I ran 12 km and then 3 km'))  
```

## Problem 7: Smart Subject Counter

Write a program that:
1) Prompts for a filename
2) Uses `try/except` to handle missing files
3) Reads the file line by line
4) Prints lines that start with `'Subject:'`
5) Counts:
   - total lines in file
   - total subject lines
6) Prints the percentage of lines that are subject lines

**Output should include** total lines, subject lines, and percent.

```python
def smart_subject_counter():
    filename = input("Enter filename: ")
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            total_lines = 0
            subject_lines = 0
            
            for line in file:
                total_lines += 1
                if line.startswith('Subject:'):
                    print(line.strip())  
                    subject_lines += 1
                    
            print("\n--- Summary ---")
            print(f"Total lines: {total_lines}")
            print(f"Subject lines: {subject_lines}")
            
            if total_lines > 0:
                percent = (subject_lines / total_lines) * 100
                print(f"Percentage: {percent:.2f}%")
            else:
                print("Percentage: 0.00%")
                
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")

smart_subject_counter()
```

## Problem 8: File Statistics Summary

Write a program that:
1) Prompts for a filename (with `try/except`)
2) Reads line by line
3) Computes:
   - total lines
   - total characters (include spaces)
   - total words (split on whitespace)
   - average words per line (non-empty lines only)
4) Prints a neat summary.

```python
def file_statistics_summary():
    filename = input("Enter filename: ")
    try:
        with open(filename, 'r', encoding='utf-8') as file:
            total_lines = 0
            total_characters = 0
            total_words = 0
            non_empty_lines = 0
            
            for line in file:
                total_lines += 1
                total_characters += len(line) 
                
                words = line.split()
                total_words += len(words)
                
                if len(words) > 0:
                    non_empty_lines += 1
                    
            print("\n--- File Statistics Summary ---")
            print(f"Total lines: {total_lines}")
            print(f"Total characters: {total_characters}")
            print(f"Total words: {total_words}")
            
            if non_empty_lines > 0:
                avg_words = total_words / non_empty_lines
                print(f"Average words per line (non-empty): {avg_words:.2f}")
            else:
                print("Average words per line (non-empty): 0.00")
                
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")

file_statistics_summary()
```
