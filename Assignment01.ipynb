## Exercise 10:

Task:
- If hours > 40, extra hours are paid at **1.5× rate**
- Otherwise normal pay  
Print: `Pay: <value>`


````python
# TODO: Write your solution here
hours = float(input('Enter hours:'))
rate = float(input('Enter rate:'))

if hours > 40:
    regular_pay = 40 * rate                 
    extra_hours = hours - 40               
    overtime_pay = extra_hours * (rate * 1.5) 
    gross_pay = regular_pay + overtime_pay
else:
    gross_pay = hours * rate
    
print('Pay: ',gross_pay)
````

## Exercise 16: Safe Pay Calculator

Task: Combine overtime + safe input:
- Ask for hours and rate
- If either is invalid: print `Error, please enter numeric input`
- Else compute overtime pay and print it


````python
# TODO: Write your solution here
hours_input = input("Enter hours: ")
rate_input = input("Enter rate: ")

try:
    hours = float(hours_input)
    rate = float(rate_input)
    
    if hours > 40:
        reg = 40 * rate
        otp = (hours - 40) * (rate * 1.5)
        pay = reg + otp
    else:
        pay = hours * rate

    print("Pay:", pay)

except:
    print("Error, please enter numeric input")
````

## Exercise 20: 

Task: Make a program that:
1. Takes user name (string)
2. Takes hours + rate (numbers, safe input using try/except)
3. Computes overtime pay
4. Prints a clean payslip


````python
# TODO: Write your solution here

user_name = input("Enter employee name: ")
hours_input = input("Enter hours worked: ")
rate_input = input("Enter hourly rate: ")

try:
    hours = float(hours_input)
    rate = float(rate_input)
    
    if hours > 40:
        regular_pay = 40 * rate
        overtime_pay = (hours - 40) * (rate * 1.5)
        total_pay = regular_pay + overtime_pay
    else:
        total_pay = hours * rate

    print(f"Name : {user_name}")
    print(f"Hours: {hours}")
    print(f"Rate : {rate}")
    print(f"Total: {total_pay}")

except ValueError:
    print("Error: Please enter valid numbers for hours and rate.")
````
