## Problem 4 — Title-Case Without `.title()`
Write a function `simple_title_case(sentence)` that:
- Strips extra spaces at the ends
- Splits words by spaces (assume single spaces between words after stripping)
- Capitalizes the first letter of each word and makes the rest lowercase
- Returns the new sentence

Example: `'hELLo wORLd'` → `'Hello World'`

````python
# Write your solution here
def simple_title_case(sentence): 
    if sentence =='' or sentence == ' ':
        return  '>  Enter a word:' 
    splited_word = sentence.split(' ')

    frist_word = splited_word[0][0].upper() + splited_word[0][1:].lower()
    second_word = splited_word[1][0].upper() + splited_word[1][1:].lower()
    return f'{frist_word} {second_word}'
   
simple_title_case('hELLo wORLd')
````

## Problem 5 — Mask Digits
Write a function `mask_digits(s)` that replaces every digit (`0`–`9`) with `*` and returns the result.

Example: `'Room 12B, Floor 7'` → `'Room **B, Floor *'`

````python
# Write your solution here

def mask_digits(s):
    for digit in '0123456789':
        s = s.replace(digit, '*')
    return s
print(mask_digits('Room 12B, Floor 7'))
````

## Problem 6 — Parse and Sum Integers from a Mixed String
Write a function `sum_numbers_in_text(text)` that finds all **positive integers** embedded in `text` and returns their sum.

Rules:
- Numbers are sequences of digits.
- Use only string operations/loops.

Example: `'I ran 12 km and then 3 km'` → `15`

````python
# Write your solution here
def sum_numbers_in_text(text):
    clean_text = ""
    for char in text:
        
        if char.isdigit():
            clean_text += char
        else:
            clean_text += " "

    total_sum = 0
    for num_str in clean_text.split():
        total_sum += int(num_str)
        
    return total_sum
print(sum_numbers_in_text('I ran 12 km and then 3 km'))
````

## **Problem 7**

Given the list:

```python
values = [12, 5, 8, 21, 3, 18, 10, 7, 16]
```

Create three new lists:

- numbers **less than 10**
- **even numbers**
- **odd numbers**

Print all three lists.

### Example Output
```python
Less than 10: [5, 8, 3, 7]
Even numbers: [12, 8, 18, 10, 16]
Odd numbers: [5, 21, 3, 7]
```

````python
# Write your solution here

values = [12, 5, 8, 21, 3, 18, 10, 7, 16]
less_than_10 =[]
evens =[]
odds =[]
for value in values:
    if value < 10:
        less_than_10.append(value)
    
    if value % 2 == 0:
        evens.append(value)
    
    if value % 2 != 0:
        odds.append(value)
print(f'less than 10 is {less_than_10}')
print(f'even is {evens}')
print(f'odd is {odds}')
````

## **Problem 9**

Given the following text:

```python
text = '''
Learning Python is fun and powerful
Python helps solve problems with simple code
Good programmers practice every day
Practice improves problem solving skills
Python makes programming enjoyable
'''
```

Write a program that:

1. Splits the text into individual words.
2. Creates a dictionary that counts how many times each word appears.
3. Print the dictionary

````python
# Write your solution here
text = '''
Learning Python is fun and powerful
Python helps solve problems with simple code
Good programmers practice every day
Practice improves problem solving skills
Python makes programming enjoyable
'''
splited_text = text.split()
count = {}
for word in splited_text:
    if word in count:
        count[word] += 1
    else:
        count[word] = 1
print(count)
````

## **Problem 11**

Given the dictionary:

```python
prices = {
    'pen': 12,
    'book': 55,
    'bag': 320,
    'eraser': 8
}
```

Convert the dictionary into a list of tuples in the form:

```python
(value, key)
```

Sort the list in **reverse order** and print it.

### Example Output
```python
[(320, 'bag'), (55, 'book'), (12, 'pen'), (8, 'eraser')]
```

````python
# Write your solution here

prices = {
    'pen': 12,
    'book': 55,
    'bag': 320,
    'eraser': 8
}
final_tup=[]

for key,value in prices.items():
    tup =(value,key)
    final_tup.append(tup)
final_tup.sort(reverse=True)
print(final_tup)
````
